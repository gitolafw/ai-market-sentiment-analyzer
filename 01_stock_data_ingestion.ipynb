```python
import yfinance as yf
import pandas as pd 
from sqlalchemy import create_engine
```

```python
tickers = ['NVDA', 'AAPL', 'MSFT', 'GOOGL', 'META', 'AMZN', 'TSLA']
```

```python
print("Downloading data...")
data = yf.download(tickers, start="2018-01-01", end="2020-12-31", interval="1d")
```

```python
df_flat = data.stack(level=1).reset_index()

df_flat['Date'] = pd.to_datetime(df_flat['Date'], format='mixed', utc=True).dt.date
```

```python
engine = create_engine('sqlite:///market_data.db')

df_flat.to_sql('stock_prices', con=engine, if_exists='replace', index=False)

print("Success! Full annual stock market data has been saved in the 'stock_prices' table.")
print(df_flat.head())
```
